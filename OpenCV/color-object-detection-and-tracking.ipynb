# Real Time Motion Detection
---

```python
# import the OpenCV library
import cv2
```

```python
# create a connection with the webcam (0 is the default camera)
camera = cv2.VideoCapture(0)

# check whether the camera is opened successfully
if not camera.isOpened():
    print("Error: Could not open webcam.")
    exit()
```

```python
previous_frame = None

# start an infinite loop
while True:

    # read one frame from the webcam 
    success, frame = camera.read()

    # if no frame was captured, stop the program
    if not success:
        print("Error: Unable to read frame from webcam.")
        break

    # convert the captured frame from BGR to Grayscale
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # blur the frame (coming in next block)
    gray_frame = cv2.GaussianBlur(gray_frame,(21, 21), 0)

    # handle the first frame
    if previous_frame is None:
        previous_frame = gray_frame
        continue

    # compare previous frame with current frame
    frame_difference = cv2.absdiff(previous_frame, gray_frame)

    _, threshold = cv2.threshold(
        frame_difference,
        25,
        255,
        cv2.THRESH_BINARY
    )

    contours, _ = cv2.findContours(
        threshold,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE
    )

    for contour in contours:

        if cv2.contourArea(contour) < 1000:
            continue

        x, y, w, h = cv2.boundingRect(contour)

        cv2.rectangle(
            frame,(x, y),(x + w, y + h),
            (0, 255, 0),2
        )

        cv2.putText(
            frame,
            "Motion Detected",(10, 30),
            cv2.FONT_HERSHEY_SIMPLEX,1,
            (0, 0, 255),2
        )

    # update the previous frame
    previous_frame = gray_frame

    # Display the captured frame in a window
    cv2.imshow("Motion Detection", frame)

    # exit the loop when 'q' is pressed.
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
```

```python
# Release the webcam so other applications can use it.
camera.release()

# Close all OpenCV windows.
cv2.destroyAllWindows()
```
