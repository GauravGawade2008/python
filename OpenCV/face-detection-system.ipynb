# Face Detection System
---

```python
# import the OpenCV library
import cv2

# Load the pre-trained Haar Cascade model.
face_detector = cv2.CascadeClassifier(
    cv2.data.haarcascades + "Haarcascade_frontalface_default.xml"
)
```

```python
# create a connection with the webcam (0 is the default camera)
camera = cv2.VideoCapture(0)

# check whether the camera is opened successfully
if not camera.isOpened():
    print("Error: Could not open webcam.")
    exit()
```

```python
# start an infinite loop
while True:

    # read one frame from the webcam 
    success, frame = camera.read()

    #if no frame was captured, stop the program
    if not success:
        print("Error: Unable to read frame from webcam.")
        break
    # convert the captured frame from BGR to Grayscale
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # The function returns the coordinates of all detected faces.
    faces = face_detector.detectMultiScale(
        gray_frame,
        scaleFactor = 1.1,
        minNeighbors = 5
    )

    for (x,y,w,h) in faces:
        cv2.rectangle(
            img=frame,
            pt1=(x, y),
            pt2=(x + w, y + h),
            color=(0, 255, 0),
            thickness=2
        )

    # count the total number of detected faces
    face_count = len(faces)

    # display the face count on the video frame
    cv2.putText(
        frame,
        f"Face Detected: {face_count}",
        (10, 30),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0,255,0),
        2
    )

    # Display the captured frame in a window
    cv2.imshow("Live Camera", frame)

    # exit the loop when 'q' is pressed.
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
```

```python
# Release the webcam so other applications can use it.
camera.release()

# Close all OpenCV windows.
cv2.destroyAllWindows()
```
