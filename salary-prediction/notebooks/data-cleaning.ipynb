---

 for model training the data is cleaned but to handle those categorical value is necessary for linear regression model 
 so those data must be scaled, and encoded for the model training
 

```python
df.groupby("company_size")["salary"].agg(["mean", "median", "count"])
```

```python
# one-hot encoding operation on categorical data
df_clean = pd.get_dummies(
    df,columns=["job_title","industry","location",'remote_work'],drop_first=True, dtype=int
)

# Ordinal Encoding operation on categorical data  
education_map = {
    "High School": 0,
    "Diploma": 1,
    "Bachelor": 2,
    "Master": 3,
    "PhD": 4
}

company_map = {
    "Startup": 0,
    "Small": 1,
    "Medium": 2,
    "Large": 3,
    "Enterprise": 4
}

df_clean["education_level"] = df_clean["education_level"].map(education_map)
df_clean["company_size"] = df_clean["company_size"].map(company_map)
```
