```python
# importing the required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# reading the csv file from the datasets 
df = pd.read_csv('..\\datasets\\job_salary_prediction_dataset.csv')
```

```python
df.head()
```

```python
df.insert(10, 'log-salary',value= np.log1p(df['salary']))
```

```python
sns.set_theme(style='whitegrid')
fig, axes = plt.subplots(1, 2, figsize= (14, 5))

sns.histplot(df['log-salary'], kde = True, ax= axes[0], color= 'blue')
axes[0].set_title('Distribution of Log Salaries')

sns.histplot(df['salary'], kde=True, ax=axes[1], color='green')
axes[1].set_title('Distribution of Salaries')

plt.tight_layout()
plt.show()
```

```python
# Median sorted categorical plot
order_jobs = df.groupby('job_title')['salary'].median().sort_values(ascending=False).index
sns.boxplot(data=df, x='job_title', hue='job_title', y='salary', order=order_jobs, )
plt.xticks(rotation=30, ha='right')
plt.show()
```

```python
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

plt.scatterplot(data=df, x='experience_years', y='salary',  ax=axes[0], color='teal', edgecolor=None)
axes[0].set_title('Scatterplot: Experience Years vs Salary', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Experience Years', fontsize=12)
axes[0].set_ylabel('Salary ', fontsize=12)

sns.scatterplot(data=df, x='skills_count', y='salary', color='purple', edgecolor=None)
axes[1].set_title('Scatterplot: Skills Count vs Salary', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Skills Count', fontsize=12)
axes[1].set_ylabel('Salary ', fontsize=12)

plt.tight_layout()
plt.show()
```
