```python
# importing the required libraries  
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# reading the csv file
df = pd.read_csv('..\\datasets\\diabetes_cleaned.csv')
print(df.head())
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df.drop(columns=['Outcome']))
plt.xticks(rotation=45)
plt.show()
```

```python
df.drop(columns=['Outcome']).hist(figsize=(15, 12), bins=20, edgecolor='black')
plt.suptitle("Feature Distributions (Post-Cleaning)", fontsize=16, y=0.93)
plt.show()
```

```python
print(df.drop(columns=['Outcome']).skew())
```

```python
plt.figure(figsize=(10, 8))

# Calculate Pearson correlation matrix
corr_matrix = df.corr()

# Plot the heatmap
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title("Correlation Matrix Heatmap", fontsize=14)
plt.show()
```
