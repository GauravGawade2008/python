---

```python
# removing the Student_ID column
df = df.drop(columns=["Student_ID"])
```

```python
df["Gender"] = df["Gender"].map({'Male': 0, 'Female': 1})
```

```python
print(df['Parental_Education'].value_counts(),'\n')
print(df['Performance_Level'].value_counts(),'\n')
```

```python
education_mapping = {
    "High School": 0,
    "Graduate": 1,
    "Postgraduate": 2
}

df["Parental_Education"] = df["Parental_Education"].map(education_mapping)
```

```python
performance_mapping = {
    "Poor": 0,
    "Average": 1,
    "Good": 2,
    "Excellent": 3
}

df["Performance_Level"] = df["Performance_Level"].map(performance_mapping)
```
